```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('cie_see.csv')
df
```

```python
x=df[['CIE']]
x
```

```python
y=df[['SEE']]
y
```

```python
sns.scatterplot(data=df,x='CIE',y='SEE')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=40)
print("Training set size:",x_train.shape[0])
print("Testining set size:",x_test.shape[0])
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
y_train_pred = lr.predict(x_train)
print("Predictions on training set:")
print(y_train_pred)
print("Predictions on testing set:")
print(y_pred)

```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mse=mean_squared_error('y_test','y_pred')
max=mean_absolute_error('y_test','y_pred')
r2=r2_score('y_test','y_pred')
print('mean_squared_error:',mse)
print('mean_absolute_error:',mse)
print('R_squared_score:',r2)
```
